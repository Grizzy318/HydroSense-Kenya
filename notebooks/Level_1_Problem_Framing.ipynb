## Problem Statement
HydroSense-Kenya aims to support efficient irrigation decisions using weather observations, soil sensor measurements, and crop parameters. The objective is to estimate crop water requirements, monitor soil moisture conditions, and improve water-use efficiency through data-driven analysis. Water scarcity and unpredictable rainfall patterns can lead to under-irrigation or over-irrigation, reducing crop yields and wasting resources. This notebook introduces the datasets, defines the core mathematical models, and establishes assumptions used throughout the project.

## Evapotranspiration (ET) Function

```python
def evapotranspiration(T, W, Solar, H):
    return max(0, 0.12*T + 0.35*W + 2.4*Solar - 0.025*H)
```

## Water Balance Function

```python
def water_balance(S, R, I, ET, D):
    return S + R + I - ET - D
```
