## Project Overview
HydroSense-Kenya is a smart irrigation decision-support system designed to improve water-use efficiency using weather, soil sensor, and crop parameter data.

## Problem Statement (500–700 words)

Agriculture in Kenya faces challenges arising from irregular rainfall patterns, increasing water scarcity, and inefficient irrigation practices. Farmers often make irrigation decisions based on experience rather than data-driven analysis, leading to either under-irrigation or over-irrigation. Under-irrigation reduces crop yields, while over-irrigation wastes water resources and may damage crops through waterlogging. The HydroSense-Kenya project aims to address these challenges by integrating weather observations, soil sensor measurements, and crop-specific parameters into a unified analytical framework.

The project focuses on estimating crop water requirements and supporting irrigation scheduling through numerical and data science techniques. Weather variables such as rainfall, temperature, humidity, solar radiation, and wind speed influence evapotranspiration, which represents the loss of water from soil and plant surfaces. Soil moisture sensors provide real-time information regarding the amount of water available in the root zone. Crop parameters describe the characteristics and water needs of different crops and farming zones. Combining these datasets enables more informed irrigation decisions.

A major objective is to model the water balance within an agricultural system. Water enters the system through rainfall and irrigation and leaves through evapotranspiration and drainage. By quantifying these flows, it becomes possible to estimate when irrigation is required and how much water should be applied. The project therefore contributes to sustainable agriculture by helping farmers conserve water while maintaining crop productivity.

The datasets used in this project contain weather measurements, soil sensor readings, and crop zone parameters. Before any modelling can be performed, the datasets must be explored and understood. Data quality issues such as missing values, duplicates, inconsistent formats, and outliers can affect the reliability of results. Consequently, a structured data analysis process is required.

The HydroSense-Kenya solution combines data analytics, numerical methods, simulation, and optimization. The first stage of the project focuses on understanding the problem domain, inspecting the datasets, and defining mathematical models that will be used throughout the remaining levels. These models include evapotranspiration estimation and water balance calculations. The insights generated form the foundation for later tasks involving vectorization, numerical methods, simulation, optimization, validation, and reproducibility.

The expected outcome is a decision-support framework capable of assisting irrigation planning and improving agricultural water management. By leveraging data-driven methods, HydroSense-Kenya seeks to provide a practical example of how computational techniques can support sustainable farming practices.

## Evapotranspiration Function

```python
def evapotranspiration(T, W, Solar, H):
    return max(0, 0.12*T + 0.35*W + 2.4*Solar - 0.025*H)
```

## Water Balance Function

```python
def water_balance(S, R, I, ET, D):
    return S + R + I - ET - D
```
