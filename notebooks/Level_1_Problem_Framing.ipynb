**10. Exploratory Data Analysis**

**Rainfall Trend**

```python
plt.figure(figsize=(10,5))
plt.plot(weather["date"], weather["rainfall_mm"])
plt.title("Daily Rainfall Trend")
plt.show()
```

**Interpretation:**

Rainfall variability influences irrigation demand. Lower rainfall periods increase dependence on artificial irrigation.

**Temperature Trend**

```python
plt.figure(figsize=(10,5))
plt.plot(weather["date"], weather["temperature_c"])
plt.show()
```

**Interpretation:**

Higher temperatures generally increase evapotranspiration and crop water demand.

**Soil Moisture by Zone**

```python
sns.boxplot(
    data=soil,
    x="zone_id",
    y="soil_moisture_pct"
)
```

**Interpretation:**

Variations indicate different irrigation requirements among zones.

**Tank Level Distribution**

```python
sns.histplot(
    soil["tank_level_liters"],
    kde=True
)
```
