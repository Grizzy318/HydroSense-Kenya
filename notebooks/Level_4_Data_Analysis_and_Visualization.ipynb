**8. Outlier Analysis
Soil Moisture Distribution**

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    y=soil["soil_moisture_pct"]
)

plt.title(
    "Soil Moisture Distribution"
)

plt.show()
```

**Interpretation**

Boxplots help identify:

* Extreme moisture readings
* Potential sensor anomalies
* Irrigation inconsistencies

**Pump Power Consumption**

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    y=soil["pump_power_watts"]
)

plt.title(
    "Pump Power Consumption"
)

plt.show()
```

**Interpretation**

Extreme values may indicate:

* High irrigation demand
* Equipment issues
* Operational inefficiencies

**9. Visualization 1: Rainfall Trend**

```python
plt.figure(figsize=(12,5))

plt.plot(
    weather["date"],
    weather["rainfall_mm"]
)

plt.xticks(rotation=45)

plt.title(
    "Daily Rainfall Trend"
)

plt.ylabel("Rainfall (mm)")

plt.show()
```

**Interpretation**

This graph shows how rainfall changes over time.

Periods of low rainfall may correspond to increased irrigation requirements.

**10. Visualization 2: Temperature Trend**

```python
plt.figure(figsize=(12,5))

plt.plot(
    weather["date"],
    weather["temperature_c"]
)

plt.xticks(rotation=45)

plt.title(
    "Daily Temperature Trend"
)

plt.ylabel("Temperature (°C)")

plt.show()
```

**Interpretation**

Higher temperatures generally increase evapotranspiration rates and crop water demand.

**11. Visualization 3: Soil Moisture by Zone**

```python
plt.figure(figsize=(8,5))

sns.boxplot(
    data=soil,
    x="zone_id",
    y="soil_moisture_pct"
)

plt.title(
    "Soil Moisture by Irrigation Zone"
)

plt.show()
```

**Interpretation**

This visualization compares moisture levels across irrigation zones.

Differences may indicate varying irrigation requirements or soil characteristics.

**12. Visualization 4: Tank Level Distribution**

```python
plt.figure(figsize=(8,5))

sns.histplot(
    soil["tank_level_liters"],
    bins=10,
    kde=True
)

plt.title(
    "Tank Water Level Distribution"
)

plt.show()
```

**Interpretation**

Tank level distributions help assess water availability and storage patterns.

**13. Visualization 5: Pump Flow Rate Distribution**

```python
plt.figure(figsize=(8,5))

sns.histplot(
    soil["pump_flow_lpm"],
    bins=10,
    kde=True
)

plt.title(
    "Pump Flow Rate Distribution"
)

plt.show()
```

**Interpretation**

Flow rate patterns indicate how irrigation water is delivered across the system.

**14. Correlation Analysis**

**Weather Variables**

```python
weather_corr = weather.select_dtypes(
    include=np.number
).corr()

weather_corr
```

**Correlation Heatmap**

```python
plt.figure(figsize=(8,6))

sns.heatmap(
    weather_corr,
    annot=True,
    cmap="coolwarm"
)

plt.title(
    "Weather Variable Correlations"
)

plt.show()
```

**Interpretation**

Correlation coefficients help identify relationships between variables.

Examples:

* Temperature and humidity
* Rainfall and solar index
* Wind speed and temperature

Strong relationships may influence irrigation planning.
