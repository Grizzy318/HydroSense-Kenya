```python
result = []

for x in temperature:
    result.append(x * 0.12)
```

**7. Evapotranspiration Function**

The project uses the simplified evapotranspiration model:

ET=max(0,0.12T+0.35W+2.4S−0.025H)

Where:

*   T = Temperature (°C)
*   W = Wind Speed (m/s)
*   S = Solar Index
*   H = Humidity (%)

**Loop-Based Implementation**

```python
def et_loop(temp, wind, solar, humidity):

    results = []

    for i in range(len(temp)):

        et = max(
            0,
            0.12*temp[i]
            + 0.35*wind[i]
            + 2.4*solar[i]
            - 0.025*humidity[i]
        )

        results.append(et)

    return np.array(results)
```

**Vectorized Implementation**

```python
def et_vectorized(temp, wind, solar, humidity):

    return np.maximum(
        0,
        0.12*temp
        + 0.35*wind
        + 2.4*solar
        - 0.025*humidity
    )
```

**8. Extract Variables**

```python
temperature = weather["temperature_c"].values
humidity = weather["humidity_pct"].values
wind = weather["wind_speed_mps"].values
solar = weather["solar_index"].values
```

**9. Benchmarking Performance**

Loop Execution

```python
start = time.time()

et1 = et_loop(
    temperature,
    wind,
    solar,
    humidity
)

loop_time = time.time() - start

print(loop_time)
```

**Vectorize Execution**

```python
start = time.time()

et2 = et_vectorized(
    temperature,
    wind,
    solar,
    humidity
)

vector_time = time.time() - start

print(vector_time)
```

**10. Performance Comparison**

```python
comparison = pd.DataFrame({

    "Method":[
        "Loop",
        "Vectorized"
    ],

    "Execution_Time":[
        loop_time,
        vector_time
    ]

})

comparison
```

Visualization

```python
plt.figure(figsize=(8,5))

plt.bar(
    comparison["Method"],
    comparison["Execution_Time"]
)

plt.title("Loop vs Vectorized Performance")

plt.ylabel("Time (seconds)")

plt.show()
```

**11. Interpretation**

The vectorized implementation should execute faster because NumPy performs calculations internally using optimized low-level code.

For larger datasets, the performance difference becomes much more significant.

This makes vectorization highly suitable for irrigation analytics systems that continuously process sensor data.

**13. Error Propagation Analysis**

Suppose temperature measurements contain an uncertainty of ±0.5°C.

```python
temperature_error = 0.5
```

**Nominal ET**

```python
et_nominal = et_vectorized(
    temperature,
    wind,
    solar,
    humidity
)
```

**High Estimate**

```python
et_high = et_vectorized(
    temperature + temperature_error,
    wind,
    solar,
    humidity
)
```

**Low Estimate**

```python
et_low = et_vectorized(
    temperature - temperature_error,
    wind,
    solar,
    humidity
)
```

**Comparison Table**

```python
error_df = pd.DataFrame({

    "Nominal_ET": et_nominal,

    "High_ET": et_high,

    "Low_ET": et_low

})

error_df.head()
```

**14. Visuaization of Errors**

```python
plt.figure(figsize=(10,5))

plt.plot(et_nominal,label="Nominal")

plt.plot(et_high,label="High")

plt.plot(et_low,label="Low")

plt.legend()

plt.title("Error Propagation in ET")

plt.show()
```
