# Level 2: NumPy Vectorization and Error Analysis

This notebook compares loop-based and vectorized computations for evapotranspiration (ET) estimation and investigates floating-point error propagation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import time
```

## Load Dataset

```python
weather = pd.read_csv('weather_daily_cleaned.csv')
weather.head()
```

## Loop-Based ET Calculation

```python
# Replace column names if necessary
T = weather.iloc[:,1].values
results=[]
start=time.time()
for t in T:
    et = 0.12*t
    results.append(et)
loop_time=time.time()-start
print('Loop Time:',loop_time)
```

## NumPy Vectorized Calculation

```python
start=time.time()
vectorized_et = 0.12*T
vector_time=time.time()-start
print('Vectorized Time:',vector_time)
```

## Performance Comparison

```python
comparison=pd.DataFrame({'Method':['Loop','Vectorized'],'Time':[loop_time,vector_time]})
comparison
```

## Floating Point Precision Experiment

```python
x=0.1
s=0
for i in range(100):
    s+=x
print('Expected:',10)
print('Actual:',s)
print('Error:',abs(10-s))
```

## Error Propagation Example

```python
temperature=np.array([25,26,27,28,29])
error=0.5
et_nominal=0.12*temperature
et_high=0.12*(temperature+error)
et_low=0.12*(temperature-error)
pd.DataFrame({'Nominal':et_nominal,'High':et_high,'Low':et_low})
```

## Visualization

```python
plt.bar(['Loop','Vectorized'],[loop_time,vector_time])
plt.title('Performance Comparison')
plt.ylabel('Execution Time (seconds)')
plt.show()
```

## Discussion
Vectorization reduces execution time by performing operations on entire arrays rather than iterating through elements one at a time. Floating-point arithmetic introduces small numerical errors that may accumulate during large computations.
