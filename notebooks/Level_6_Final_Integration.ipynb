**8. Testing Mathematical Functions**

**Evapotranspiration Function**

```python
def evapotranspiration(
    T,
    W,
    S,
    H
):

    return max(
        0,
        0.12*T
        +0.35*W
        +2.4*S
        -0.025*H
    )
```

**Test Case 1**

```python
result = evapotranspiration(
    25,
    3,
    1.2,
    60
)

print(result)
```

**Assertion Test**

```python
assert result >= 0

print("Test Passed")
```

**9.Water Balance Validation**

**Function**

```python
def water_balance(
    S,
    R,
    I,
    ET,
    D
):

    return (
        S
        +R
        +I
        -ET
        -D
    )
```

**Test Example**

```python
balance = water_balance(
    50,
    10,
    8,
    4,
    2
)

print(balance)
```

**Validation**

```python
assert balance == 62

print("Validation Passed")
```
